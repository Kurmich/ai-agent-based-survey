```python
from utils import get_codebook
from utils import get_questions_list, get_codebook_subset, codebook_to_json
import pandas as pd
```

```python
codebook = get_codebook()
question_codes = ['SOC1',	'SOC2A', 'SOC2B', 'SOC3A',	'SOC3B',	'SOC4A',	'SOC4B',	'PHYS8',	'PHYS1A',	'PHYS1B']
idxs = [i for i in range(11, 21)]
```

```python
def generate_script(idxs, clean_df, question_codes):
    scripts = {}
    for idx in idxs: scripts[idx] = []
    for code in question_codes:
        for idx in idxs:
            qa_details = codebook[code]
            ans_id = clean_df[code][idx]
            #print(clean_df[code][idx])
            question = qa_details.get_question()
            answer   = qa_details.get_answer(ans_id)
            scripts[idx].append('Q: %s\nA: %s' %(question, answer))

    for idx in idxs: scripts[idx] = '\n'.join(scripts[idx])
    return scripts
```

```python
def get_clean_df_subset(df, idxs, question_codes):
    return df.loc[idxs, question_codes]
```

```python
clean_df = pd.read_csv('clean_april.csv')
clean_df_subset = get_clean_df_subset(clean_df, idxs, question_codes)
clean_df_subset.to_csv('clean_april_subset2.csv')
```

```python
cb_subset = get_codebook_subset(codebook, question_codes)
codebook_to_json(cb_subset)
```

```python
scripts = generate_script(idxs, clean_df_subset,question_codes )
question_list = get_questions_list(codebook, question_codes)
```
